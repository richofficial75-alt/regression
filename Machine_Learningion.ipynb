 **3.How would you handle missing values in a dataset? Explain at least three
methods with examples.                                    
(Hint: Consider deletion, mean/median imputation, and predictive modeling).**

- Handling missing values is a crucial part of data preprocessing because poor handling can lead to biased results or loss of valuable information.

**1. Deletion Methods**

 **A) Listwise Deletion**

  You remove any row that contains a missing value.

use:
- Missing values are very few
- Dataset is large
-  Missingness is random (MCAR)
- *Example- Dataset*

 **b) Column Deletion**
If a column has too many missing values (e.g., 70–80%), remove that feature.
When to use:
- Very high missing percentage
-  Feature is not important

  **2. Imputation**

  **A) Mean/Median Imputation**
Replace missing values with the mean (for normal distribution) or median (for skewed data).

- Example:
Age column: [25, 30, NaN, 40]
Mean = (25 + 30 + 40) / 3 = 31.67
- Imputed column:
[25, 30, 31.67, 40]

  **B) Mode Imputation**

   Example:

   City: ["Delhi", "Mumbai", NaN, "Delhi"]
   
   Mode = "Delhi"

 Imputed: ["Delhi", "Mumbai", "Delhi", "Delhi"]

 **3. Predictive Modeling**
 - Use ML models to predict missing values based on other features.

  **a) Regression Imputation**

 - Use a regression model to predict the missing numerical value.
 - Example:
Predict missing Salary using features like Age, Education, Experience.

 **b) KNN Imputation (K-Nearest Neighbors)**

 Find similar rows and fill missing values using the neighbors’ averages.
 - Example:
If Age is missing, KNN finds similar rows and imputes based on nearest values.

  **c) Multiple Imputation**

Creates multiple predictions and averages them for more accuracy.

  use:
 - Missing values are not random
 - Important features cannot be deleted
 - You want higher accuracy

**5. Why is feature scaling important in ML? Compare Min-Max scaling and Standardization.**

**Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent.**
- Feature scaling is important in machine learning because many algorithms perform better and faster when all features are on a similar scale.

 **Compare Min-Max scaling and Standardization**

| Feature               | Min-Max Scaling  | Standardization          |
| --------------------- | ---------------- | ------------------------ |
| Range                 | 0 to 1           | Mean = 0, SD = 1         |
| Sensitive to Outliers | Yes              | Less                     |
| Suitable For          | NN, KNN, K-means | SVM, LR, PCA             |
| Data Distribution     | Any              | Best when approx. normal |
| Bounded Output        | Yes              | No                      
 |

**Impact on K-Nearest Neighbors (KNN)**
KNN classifies a point based on the distance between data points:


If one feature has a large scale (e.g., salary = 0–1,00,000), it will dominate another feature with small scale (e.g., age = 18–60).

Impact:

- Biases the distance calculation

- Important small-range features become irrelevant

- Wrong neighbors are selected

- Poor accuracy

**Support Vector Machines (SVM)**

SVM tries to find the maximum-margin hyperplane.
Margin calculation depends on:

- If one feature has a large magnitude:

- SVM gives larger weights to smaller-scale features
  becomes skewed

- Hyperplane is incorrectly oriented

Impact:

- Incorrect decision boundary

- Longer training time

- Poor classification

- Kernel functions (RBF kernel) become distorted

**Impact on Gradient Descent**

 - In neural networks or linear models using gradient descent:
Large-scale features cause the cost function to have a steep and uneven surface, slowing down convergence.

**6. Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other?**

**Hint: Consider categorical variables with ordinal vs. nominal relationships.**

**Label Encoding**
Assigns each category a unique integer.
Example:
{'Red': 0, 'Blue': 1, 'Green': 2}
 *Advantages*
 - Simple and fast
 - Does not increase feature dimensions
 - Useful when categories have a natural order.

 **One-Hot Encoding**

Creates a new binary column for each category (0/1).
Example for Red, Blue, Green:

*Advantages*
- No ordinal relationship assumed
- Works well for nominal categories
- Prevents algorithms from misinterpreting category ranks.
**Use Label Encoding**
- The categories are ordinal (have natural order).
 **Examples**
 - Education level → Primary < Secondary < Graduate
 - Size → Small < Medium < Large
 - Ratings → Bad < Average < Good
 - Tree-based models
can still work well with label-encoded features because they don’t rely on distances.
**Use One-Hot Encoding**

The categories are nominal.
**Examples**
 - Colors → Red, Blue, Green
 - City → Mumbai, Delhi, Kolkata
 - Gender → Male, Female, Other
