1. **What is Simple Linear Regression (SLR)? Explain its purpose?**

  - A statistical and machine learning technique used to model the relationship between two continuous quantitative variables by fitting a straight line through observed data called Simple Linear Regression.
  - **Purpose** - Predictive Modeling, Trend Identification, Explanatory Analysis, Optimization: In business or scientific contexts, it helps determine the optimal settings for a factor to achieve a target outcome.

2. **What are the key assumptions of Simple Linear Regression?**

- **Linearity-** The relationship between the independent variable (\(X\)) and the dependent variable (\(Y\)) must be linear.

 - **Independence of Errors**- The residuals (differences between actual and predicted values) for different observations must be independent of each other. This is particularly critical in time-series data, where the error at one point in time should not predict the error at the next.

 - **Normality of Residuals**- The residuals of the model should follow a normal distribution (a bell-shaped curve centered at zero). This assumption is necessary for validly performing hypothesis tests and calculating confidence intervals.

-  **No Significant Outliers**- SLR is highly sensitive to outliers, which are data points that lie far from the general pattern. These points can disproportionately pull the regression line away from the bulk of the data, skewing the results.
- **Homoscedasticity**- The variance of the residuals should be constant across all levels of the independent variable. If the error spread increases or decreases (e.g., forming a "fan" or "cone" shape), the data is said to have heteroscedasticity, which makes standard errors unreliable.

**5.** **What is the method of least squares in linear regression?**
- In linear regression, the method of least squares is a statistical technique used to determine the "best-fit" line for a set of data points. It provides a mathematical way to find a relationship between an independent variable (x) and a dependent variable (y)
- The "best fit" is defined as the line that minimizes the sum of the squares of the vertical differences (residuals) between the observed data points and the values predicted by the model.

**7. Name and briefly describe three common evaluation metrics for regression models.**
- Three common evaluation metrics for regression models are as follows.
- **Mean Absolute Error (MAE)**: This metric calculates the average of the absolute differences between the actual values and the model's predictions. It provides a straightforward measure of error magnitude in the same units as the target variable.
- **Root Mean Squared Error (RMSE)**: RMSE is the square root of the average of squared differences between actual and predicted values. Like MAE, it is expressed in the same units as the target variable, making it highly interpretable for stakeholders. However, because it squares the errors before averaging, it penalizes large errors more heavily, making it useful for applications where significant deviations are particularly costly.
- **R-Squared**: Also known as the coefficient of determination, this metric measures the proportion of variance in the dependent variable that is explained by the independent variables. It typically ranges from 0 to 1 (or 0% to 100%), where 1 indicates a perfect fit and 0 indicates the model performs no better than simply predicting the mean of the data.

**8. What is the purpose of the R-squared metric in regression analysis?**

- In regression analysis, the primary purpose of R-squared, also known as the coefficient of determination, is to measure the "goodness-of-fit" of a model. It quantifies how much of the variation in a dependent variable is explained by the independent variable(s) in a model.

9. **Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept. (Include your Python code and output in the code box below**

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Prepare the data
# X must be a 2D array (reshape is used for a single feature)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5])

# 2. Create and fit the model
model = LinearRegression()
model.fit(X, y)

# 3. Retrieve and print results
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (Coefficient): {slope}")
print(f"Intercept: {intercept}")
```
